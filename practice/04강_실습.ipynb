## 원뿔의 부피&겉넓이 계산 프로그램 5


```python
# 두 줄의 코드를 중첩 등을 통해 한 줄로 간결하게 표현하면 프로그램이 더 세련되어집니다.
rad = int(input())
# input()에서 받은 값은 문자열로 처리

print(type(rad))
print(rad)
```

```python
# [[원뿔 부피 및 겉넓이 계산 프로그램]]
rad = int(input("반지름을 입력하세요: "))
hei = int(input("높이를 입력하세요: "))
sla = pow(rad ** 2 + hei ** 2, 0.5)

# 모선 계산
sla = pow(rad**2 + hei**2, 0.5)

# vol = 부피
vol = 1 / 3 * 3.14 * rad ** 2 * hei

# suf = 겉넓이
suf = 3.14 * rad ** 2 + 3.14 * rad * sla

# [부피 & 겉넓이 출력]
# print() 함수: 여러 데이터를 한 번에 출력 가능, 콤마(,)로 연결된 부분은 공백이 추가되어 출력
# sep옵션 활용하여 공백 처리하기
print("원뿔의 부피들", str(vol) + "입니다.", sep='|')

# 두 데이터의 에러없는 결합을 위해 숫자를 str()을 통해 문자열로 변환합니다.
print("원뿔의 겉넓이는", str(suf) + "입니다.")


```
