## 3강 2절 - 연산자

```python
# 2 ** 10, 5 / 2 vs 5 // 2, 25 % 7, 8.4 % 0.9, number % 2
print(2 ** 10)
print(5 / 2)
print(5 // 2)
print (8.4 % 0.9) # 모듈로, 근사치 소수들로 모률로 연산을 했기 때문에 오차가 발생
```

```python
# max(2, 3, 4), min(2, 3, 4), round(3.141592, 2), round(3.415), abs(-3), pow(2, 3)
print(pow (2, 3))
print(pow (9, 0.5))
```

```python
rad = 10
hei = 24
sla = pow(rad ** 2 + hei ** 2, 0.5)
print(1 / 3 * 3.14 * rad ** 2 * hei)
print(3.14 * rad ** 2 + 3.14 * rad * sla)
```

## 3강 3절 - 파이썬 기초문법 상식

```python
# 평균학점 계산: 계산 방법 sum(수강점수) / 이수학점
print((4.5 * 2 + 4.0 * 3) / (130 - 66))
```

```python
#원뿔의 높이와 반지름을 활용하여 부피와 겉넓이를 계산하는 프로그램
# rad = 반지름, hei = 높이
rad = 10
hei = 24
# 모선 계산
sla = pow(rad ** 2 + hei ** 2, 0.5)
# 부피 출력
print(1 / 3 * 3.14 * rad ** 2 * hei)
print(3.14 * rad ** 2 + 3.14 * rad * sla)
```
