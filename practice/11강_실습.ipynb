**math 모듈을 사용한 원뿔 계산 프로그램**

```python
import math

# 원뿔 클래스 정의
class Cone :
    def __init__(self, radius = 20, height = 30):
        self.radius = radius
        self.height = height

    def get_vol(self) :
            return 1/3 * math.pi * self.radius ** 2 * self.height

    def get_surf (self) :
        sla = pow (self.radius ** 2 + self.height ** 2, 0.5)
        return math.pi * self.radius * (self.radius + sla)

first = Cone(50, 50)
print("큰 원뿔의 부피는", first.get_vol(), "입니다.")
print("큰 원뿔의 겉넓이는", first.get_surf(), "입니다.")
```

**로또 추첨 프로그램**

1 ~ 45 숫자 6개를 입력 받아 당첨 숫자와 비교

```python
import random
guess_str = input("1~45 번호 6개를 입력하세요: ").split(", ")

guess_list = [int(i) for i in guess_str if int(i) < 46]

if len(guess_list) == 6:
    lotto_list = random.sample(range(1, 46, 1), 6)
    print("당첨 숫자는", lotto_list, "입니다.")
    hit_count = 0

    for i in guess_list:
        for j in lotto_list:
            if i == j :
                hit_count += 1

    print(str(hit_count) + "개 맞혔습니다.")
else :
    print("45 이하의 번호 6개를 넣으세요")
```

**스무고개 프로그램**

20번의 기회 안에 1 ~ 100 사이의 숫자를 맞히기


```python
import random
hit_number = random.randint(1, 501) # 1~500까지
# print(hit_number)

guess_count_list = range(1, 21, 1) # 기회는 20번
passed = False
for guess_count in guess_count_list :
    guess = int (input("숫자를 맞혀보세요(" + str(guess_count) + "번째 시도): "))

    if hit_number == guess :
        passed = True
        break
    elif hit_number > guess :
        print(str(guess) + "보다 큽니다.", end="")
    else :
        print(str(guess) + "보다 작습니다.", end="")



if passed == True:
    print("맞혔습니다. 축하합니다.")
else:
    print("모든 기회를 다 사용하셨습니다. ")
```

**5000보다 작은 소수 찾기 (2~5000 하나하나 다 세야 함)**

```python
import time

max = 90000
def Isprime(x):
    if x < 2: # Numbers less than 2 are not prime
        return False
    for i in range(2, int(x**0.5) + 1): # Check divisibility up to sqrt(x)
        if x % i == 0:
            return False
    return True

pn_count = 0
start_time = time.time()
for i in range(2, max+1):
    if Isprime(i):
        pn_count += 1
        print(i, end=", ")
        if pn_count % 5 == 0:
            print("")


print("")
end_time = time.time()

print(format(end_time - start_time, ".3f"), "초 실행했습니다.")
```
