### 텍스트 데이터 인사이트 분석기

**데이터 전처리(데이터 변환 및 불용어 필터링)**

```python
PUNCTUATIONS = ".,!?\"':;()-"

def clean_text(fp):
    text = ""

    for line in fp:
        #소문자 변환
        line = line.lower()

        for punc in PUNCTUATIONS:
            line = line.replace(punc, "")
        text += line + " "

    return text.strip()
```

**데이터 분석(단어 카운트)**

```python
# 불용어 집합
STOPWORDS = {"a", "the", "is", "of", "in", "to", "and", "i", "it", "that", \
             "this", "my", "was", "for", "as", "but", "so", "be", "he", \
             "me", "his", "him", "she", "her", "you", "your", "we", "our", "us", \
             "on", "were", "been", "are"}

def count_words(text):
    words = text.split()
    counts = {}

    for word in words:
        word = word.strip()

        if not word: # Skip empty strings after stripping
            continue

        if word in STOPWORDS:
            continue

        counts[word] = counts.get(word, 0) + 1

    return counts
```

```python
def get_sorted_list(counts_dict):
    temp_list = []

    # Iterate over key-value pairs using .items()
    for word, count in counts_dict.items():
        data_pair = (count, word) # Store as (count, word) for sorting
        temp_list.append(data_pair)

    temp_list.sort(reverse=True)

    return temp_list

def print_save_graph(filename, sorted_list, tops = 10):

    re_fp = open(filename, "w")

    title = "== 데이터 분석결과 보고서 =="
    print(title)

    re_fp.write(title + "\n")

    rank = 1

    for item in sorted_list:

        if rank > tops:
            break

        count = item[0]
        word = item[1]

        graph =  "#" * (count // 100)
        result_line = format(rank, ">3n") + "." + format(word, "<15s") +": " + graph + "(" + str(count) + ")"
        print(result_line)


        re_fp.write(result_line + "\n")

        rank += 1

    re_fp.close()
```
