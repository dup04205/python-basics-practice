문제 2

```python
# 1. 사용자로부터 총 초를 정수형으로 입력받음
total_seconds = int(input("총 초를 입력하세요: "))

# 2. 시 계산: 3600으로 나눈 몫
hours = total_seconds // 3600

# 3. 남은 초 계산: 3600으로 나눈 나머지
remaining_seconds = total_seconds % 3600

# 4. 분 계산: 남은 초를 60으로 나눈 몫
minutes = remaining_seconds // 60

# 5. 최종 초 계산: 60으로 나눈 나머지
seconds = remaining_seconds % 60

# 6. 결과 출력
print(f"{total_seconds}초는 {hours}시간 {minutes}분 {seconds}초입니다.")
```

문제 3

```python
# 점수 입력 (실수형으로 변환하여 소수점 점수도 처리 가능하도록 구성)
score = float(input("점수를 입력하시오: "))

# 선택 구조를 통한 학점 판별
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

# 결과 출력
print(f"학점: {grade}")
```

문제 4

```python
def check_leap_year():
    try:
        year = int(input("연도를 입력하세요: "))

        # 윤년 판별 핵심 논리
        if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
            print(f"{year}년은 윤년이다.")
        else:
            print(f"{year}년은 평년이다.")

    except ValueError:
        print("유효한 정수를 입력해야 한다.")

# 프로그램 실행
check_leap_year()
```

문제 5

```python
# 구구단 출력 프로그램

# 1. 출력할 단 입력 (정수형 변환)
dan = int(input("출력할 단을 입력하세요: "))

# 2. 제어 변수 초기화
i = 1

# 3. while 반복 구조 및 조건식 설정
while i <= 9:
    # 반복 실행문: 구구단 결과 출력 (f-string 포매팅 사용)
    print(f"{dan} * {i} = {dan * i}")

    # 4. 증감식 기술: 제어 변수 1 증가
    i += 1
```
